с нормализацией пока сильно не заморачивался. Для этого надо будет для начала разбить на сплиты. Пока сделал просто мин-максом

```python
normalize_columns = ['Methane', 'Ethylene'] + [f's{i}' for i in range(1, 17)]
```

```python
def min_max_scaler(normalize_columns, ds):
    for col in normalize_columns:
        ds[col] = (ds[col] - ds[col].min()) / (ds[col].max() - ds[col].min())
```

```python
min_max_scaler(normalize_columns, ethylene_co)
```

```python
min_max_scaler(normalize_columns, ethylene_methane)
```

Попробую сделать со standard скейлером. Есть подозрение, что эти странные значения, лежащие около ноля, могут быть связаны с использованием minmax-а

```python
def standard_scaler(normalize_columns, ds):
    for col in normalize_columns:
        ds[col] = (ds[col] - ds[col].mean()) / (ds[col].std())
```

```python
standard_scaler(normalize_columns, ethylene_co)
```

```python
standard_scaler(normalize_columns, ethylene_methane)
```
